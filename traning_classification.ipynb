```python
import pandas as pd
```

```python
df = pd.read_csv("zoo_python.csv")
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label = LabelEncoder()
```

```python
df['animal']=label.fit_transform(df['animal'])
df['uniq_id']=label.fit_transform(df['uniq_id'])
df['water_need']=label.fit_transform(df['water_need'])
df['food']=label.fit_transform(df['food'])
```

```python
df
```

```python
x=df.drop('food',axis=1)
y=df['food']
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.3,random_state=0)
```

```python
from sklearn.tree import DecisionTreeClassifier
tree1 = DecisionTreeClassifier(criterion = 'gini' , max_depth=3 , random_state=0)
```

```python
tree1.fit(x_train,y_train)
```

```python
y_pred=tree1.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import classification_report
mse = classification_report(y_test,y_pred)
print(mse)
```

```python
from sklearn.metrics import classification_report , confusion_matrix , accuracy_score
```

```python
mse2 = accuracy_score(y_test,y_pred)
print(mse2)
```

```python
mse3 = confusion_matrix(y_test,y_pred)
print(mse3)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree as t
plt.figure(figsize=(19,7))
t.plot_tree(tree1.fit(x_train,y_train))
```

```python
import pandas as pd
```

```python
df = pd.read_csv("zoo_python.csv")
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label = LabelEncoder()
df['l_animal'] = label.fit_transform(df['animal'])
df['l_uniq_id'] = label.fit_transform(df['uniq_id'])
df['l_water_need'] = label.fit_transform(df['water_need'])
df['l_food'] = label.fit_transform(df['food'])
```

```python
df
```

```python
df = df.drop(['animal','uniq_id','water_need','food'],axis=1)
```

```python
df
```

```python
x=df.drop('l_food',axis=1)
y=df['l_food']
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.3,random_state=0)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tree = DecisionTreeClassifier( criterion='entropy', max_depth=3 , random_state=0)
```

```python
tree
```

```python
tree.fit(x_train,y_train)
```

```python
y_pred = tree.predict(x_test)
```

```python
from sklearn.metrics import classification_report , confusion_matrix , accuracy_score
```

```python
report = classification_report (y_test , y_pred)
```

```python
import pandas as pd 
```

```python
dataset_market_1 = [['Milk', 'Onion', 'Nutmeg', 'Kidney Beans', 'Eggs', 'Yogurt'],
           ['Dill', 'Onion', 'Nutmeg', 'Kidney Beans', 'Eggs', 'Yogurt'],
           ['Milk', 'Apple', 'Kidney Beans', 'Eggs'],
           ['Milk', 'Unicorn', 'Corn', 'Kidney Beans', 'Yogurt'],
           ['Corn', 'Onion', 'Onion', 'Kidney Beans', 'Ice cream', 'Eggs']]
```

```python
from mlxtend.preprocessing import TransactionEncoder
encoder = TransactionEncoder()
data = encoder.fit(dataset_market_1).transform(dataset_market_1).astype('int')
```

```python
data
```

```python
data = pd.DataFrame(data)
```

```python
data
```

```python
data.columns=encoder.columns_
```

```python
data
```

```python
from mlxtend.frequent_patterns import apriori
```

```python
data_assoc = apriori(data , min_support = 0.5 , use_colnames=True )
```

```python
data_assco
```

```python
from mlxtend.frequent_patterns import association_rules
```

```python
rules = association_rules(data_assoc , metric = 'confidence', min_threshold = 0.6)
```

```python
rules
```

```python
dataset_market_1 = [['Milk', 'Onion', 'Nutmeg', 'Kidney Beans', 'Eggs', 'Yogurt'],
           ['Dill', 'Onion', 'Nutmeg', 'Kidney Beans', 'Eggs', 'Yogurt'],
           ['Milk', 'Apple', 'Kidney Beans', 'Eggs'],
           ['Milk', 'Unicorn', 'Corn', 'Kidney Beans', 'Yogurt'],
           ['Corn', 'Onion', 'Onion', 'Kidney Beans', 'Ice cream', 'Eggs']]
```

```python
from mlxtend.preprocessing import TransactionEncoder
tran = TransactionEncoder()
array = tran.fit(dataset_market_1).transform(dataset_market_1).astype('int')
```

```python
array
```

```python
data = pd.DataFrame(array)
```

```python
data
```

```python
data.columns=tran.columns_
```

```python
from mlxtend.frequent_patterns import apriori , association_rules
```

```python
freq_data = apriori(data , min_support = 0.3 , use_colnames =True)
```

```python
freq_data
```

```python
ass_data = association_rules(freq_data , metric = 'confidence' , min_threshold=0.6)
```
